# 03 Monte Carlo Simulation

Estimate the probability of at least one crash in one selected ZIP/day/hour block.

```python
from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.monte_carlo import predict_for_zip_hour
```

```python
rate_path = PROJECT_ROOT / "data" / "processed" / "crash_rate_table.csv"
rate_table = pd.read_csv(rate_path, dtype={"zip_code": str})

zip_code = "10001"
day_of_week = "Friday"
hour = 17
weather_condition = "rain"

candidate = rate_table[
    (rate_table["zip_code"] == zip_code)
    & (rate_table["day_of_week"] == day_of_week)
    & (rate_table["hour"] == hour)
]
if candidate.empty:
    example_row = rate_table.iloc[0]
    zip_code = str(example_row["zip_code"])
    day_of_week = example_row["day_of_week"]
    hour = int(example_row["hour"])

result = predict_for_zip_hour(
    rate_table,
    zip_code=zip_code,
    day_of_week=day_of_week,
    hour=hour,
    weather_condition=weather_condition,
    num_trials=10000,
    random_seed=42,
)
result_summary = {key: value for key, value in result.items() if key != "simulated_counts"}
result_summary
```

The main probability is the share of Monte Carlo trials where the simulated crash count is at least one.

```python
f"P(at least one crash) = {result['probability_at_least_one']:.2%}"
```
