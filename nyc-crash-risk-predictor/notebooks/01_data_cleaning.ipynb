# 01 Data Cleaning

Download a manageable sample of NYC crash data and turn the raw fields into clean modeling columns.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.download_data import download_crash_data
from src.clean_crashes import clean_crash_data
```

```python
raw_path = PROJECT_ROOT / "data" / "raw" / "nyc_crashes_raw.csv"
cleaned_path = PROJECT_ROOT / "data" / "processed" / "nyc_crashes_clean.csv"

if not raw_path.exists():
    download_crash_data(limit=50000, output_path=raw_path)

cleaned_df = clean_crash_data(input_path=raw_path, output_path=cleaned_path)
cleaned_df.head()
```
